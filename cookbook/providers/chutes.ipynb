```python
## Setup

You'll need both a Portkey API key and a Chutes.ai API key.
```

```python
import os
from openai import OpenAI
from portkey_ai import PORTKEY_GATEWAY_URL, createHeaders

# Set your API keys
PORTKEY_API_KEY = os.environ.get("PORTKEY_API_KEY")
CHUTES_API_KEY = os.environ.get("CHUTES_API_KEY")

client = OpenAI(
    api_key=CHUTES_API_KEY,
    base_url=PORTKEY_GATEWAY_URL,
    default_headers=createHeaders(
        provider="chutes",
        api_key=PORTKEY_API_KEY
    )
)
```

```python
## List Available Models

Chutes.ai provides access to many state-of-the-art models. Let's see what's available:
```

```python
# List all available Chutes models
models = client.models.list()
print(f"Available models: {len(models.data)}")
for model in models.data[:10]:  # Show first 10 models
    print(f"- {model.id}")
```

```python
## Chat Completion

Let's use one of the models for a chat completion:
```

```python
chat_completion = client.chat.completions.create(
    model="deepseek-ai/DeepSeek-V3-0324",  # or any other model from the list
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Explain quantum computing in simple terms"}
    ],
    max_tokens=150
)

print(chat_completion.choices[0].message.content)
```
